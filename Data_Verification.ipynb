```python
from pathlib import Path
import pandas as pd
import numpy as np

AFS_DIR = Path("..") / "data" / "AFS"

print("AFS directory:", AFS_DIR.resolve())

pair_files = [
    "ArgPairs_DP.csv",
    "ArgPairs_GC.csv",
    "ArgPairs_GM.csv"
]

for file in pair_files:
    path = AFS_DIR / file

    print(f"{file}:")
    print("  Exists:", path.exists())
    print("  Size:", path.stat().st_size if path.exists() else "N/A")
    print()
```

```python
datasets = {}

for file in pair_files:
    path = AFS_DIR / file
    
    df = pd.read_csv(
        path,
        encoding="latin1"
    )
    
    datasets[file] = df
    
    print("=" * 70)
    print(file)
    print("Shape:", df.shape)
    print("Columns:")
    print(df.columns.tolist())
```

```python
for file, df in datasets.items():
    print("=" * 70)
    print(file)
    display(df.head(5))
```

```python
required_columns = [
    "regression_label",
    "sentence_1",
    "sentence_2"
]

for file, df in datasets.items():
    print("=" * 70)
    print(file)
    
    for column in required_columns:
        print(
            f"{column}:",
            "FOUND" if column in df.columns else "MISSING"
        )
```

```python
for file, df in datasets.items():
    scores = pd.to_numeric(
        df["regression_label"],
        errors="coerce"
    )
    
    print("=" * 70)
    print(file)
    print("Number of scores:", len(scores))
    print("Missing scores:", scores.isna().sum())
    print("Minimum:", scores.min())
    print("Maximum:", scores.max())
    print("Unique scores:", scores.nunique())
```

```python
for file, df in datasets.items():
    
    pairs = df[
        ["sentence_1", "sentence_2"]
    ].astype(str)
    
    duplicate_count = pairs.duplicated().sum()
    
    print("=" * 70)
    print(file)
    print("Duplicate sentence pairs:", duplicate_count)
```

```python
afs_parts = []

topic_mapping = {
    "ArgPairs_DP.csv": "death_penalty",
    "ArgPairs_GC.csv": "gun_control",
    "ArgPairs_GM.csv": "gay_marriage"
}

for file, df in datasets.items():
    temp = df[
        ["sentence_1", "sentence_2", "regression_label"]
    ].copy()
    
    temp["topic"] = topic_mapping[file]
    
    afs_parts.append(temp)

afs = pd.concat(
    afs_parts,
    ignore_index=True
)

print("Combined shape:", afs.shape)
display(afs.head())
```

```python
print("Total pairs:", len(afs))
print()

print("Pairs by topic:")
print(afs["topic"].value_counts())
print()

print("Missing values:")
print(afs.isna().sum())
print()

print("Score range:")
print(
    "Min:", afs["regression_label"].min(),
    "Max:", afs["regression_label"].max()
)
```
